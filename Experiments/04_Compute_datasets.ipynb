```python
from pathlib import Path
import pickle
import pandas as pd
import numpy as np
```

```python
df_nt = pd.read_csv('files/xt_nontop.csv')
df_ap = pd.read_csv('files/xt_alpha.csv')
df_cap = pd.read_csv('files/xt_calpha.csv')
df_sp = pd.read_csv('files/xt_sixpack.csv')
```

```python
datasets = []
```

### Non topological

```python
X = df_nt.iloc[:,3:].to_numpy()
```

```python
values = X
datasets.append(
    {
        'key': 'nontop',
        'title': 'Non-top.',
        'values': values,
    }
)
```

### Betti curves of Six-pack

```python
X = df_sp.iloc[:,3:].to_numpy()
```

```python
values = X[:,0:300]
datasets.append(
    {
        'key': 'sixpackMT0',
        'title': 'Six-pack, M vs T, dim. 0',
        'values': values,
    }
)
values = X[:,300:600]
datasets.append(
    {
        'key': 'sixpackMT1',
        'title': 'Six-pack, M vs T, dim. 1',
        'values': values,
    }
)
values = X[:,0:300] - X[:,300:600]
datasets.append(
    {
        'key': 'sixpackMTeuler',
        'title': 'Six-pack, M vs T, Euler char.',
        'values': values,
    }
)
values = X[:,600:900]
datasets.append(
    {
        'key': 'sixpackTV0',
        'title': 'Six-pack, T vs V, dim. 0',
        'values': values,
    }
)
values = X[:,900:1200]
datasets.append(
    {
        'key': 'sixpackTV1',
        'title': 'Six-pack, T vs V, dim. 1',
        'values': values,   
    }
)
values = X[:,600:900] - X[:,900:1200]
datasets.append(
    {
        'key': 'sixpackTVeuler',
        'title': 'Six-pack, T vs V, Euler char.',
        'values': values,  
    }
)
values = X[:,1200:1500]
datasets.append(
    {
        'key': 'sixpackMV0',
        'title': 'Six-pack, M vs V, dim. 0',
        'values': values,    
    }
)
values = X[:,1500:1800]
datasets.append(
    {
        'key': 'sixpackMV1',
        'title': 'Six-pack, M vs V, dim. 1',
        'values': values, 
    }
)
values = X[:,1200:1500] - X[:,1500:1800]
datasets.append(
    {
        'key': 'sixpackMVeuler',
        'title': 'Six-pack, M vs V, Euler char.',
        'values': values,  
    }
)
```

### Non-topological + Betti curves

```python
nontop_values = datasets[0]['values']
for dataset in datasets[1:]:
    combined_values = np.hstack([
        nontop_values,
        dataset['values']
    ])
    datasets.append({
        'key': f"nontop+{dataset['key']}",
        'title': f"Non-top. + {dataset['title']}",
        'values': combined_values
    })
```

```python
with open('files/datasets.pkl', 'wb') as f:
    pickle.dump(datasets, f)
```
