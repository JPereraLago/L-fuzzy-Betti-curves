```python
import numpy as np
import pandas as pd
import pickle
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import f1_score
from utils import *
```

```python
with open('files/datasets.pkl', 'rb') as f:
    datasets = pickle.load(f)
```

```python
n_iter = 100
n_trees = 100
test_size = 0.25
```

```python
df_l = pd.read_csv('files/xt.csv')
y = df_l['label'].to_numpy()
```

```python
ld = len(datasets)
for cd, d in enumerate(datasets, start=1):
    title = d['title']
    X = d['values']
    tqdm.write(f'{title} ({cd}/{ld})')
    f1_array = np.empty(n_iter)
    for i in tqdm(range(n_iter), desc=f'Progress ({cd}/{ld})'):
        X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=test_size, stratify=y, random_state=i)
        model = RandomForestClassifier(n_estimators=n_trees, random_state=2*i, class_weight="balanced")
        model.fit(X_train, y_train)
        y_pred = model.predict(X_test)
        f1_array[i] = f1_score(y_test, y_pred, average="macro")
    d['f1'] = f1_array
    with open('files/datasets.pkl', 'wb') as f:
        pickle.dump(datasets, f)
```
